```python
import pandas as pd

df = pd.read_csv("../data/cleaned.csv")
grouped_tempo = df.groupby('tempo_level').agg({
    'popularity' : 'mean',
})
print(grouped_tempo)
```

Insight: High tempo songs are the most popular. After that comes the medium tempo songs that are almost as popular. Low tempo songs are less popular. However the difference between high and medium tempo songs is insignifcant suggesting weak correlation.

```python
grouped_energy = df.groupby('energy_level').agg({
    'popularity' : 'mean',
})
print(grouped_energy)
```

Insight: Medium energy songs are generally more popular. After that comes high energy songs. Low energy songs are the least popular. Therefore, energy levels show a clear pattern where high and medium energy songs outperform low energy songs in popularity. Energy seems more important than tempo in determining popularity.

```python
grouped_popularity = df.groupby('popularity_level').agg({
    'danceability' : 'mean',
    'loudness' : 'mean',
    'acousticness' : 'mean',
    'energy' : 'mean',
    'tempo' : 'mean',
})
print(grouped_popularity)
```

Louder songs are more popular. The danceability of the songs affects their popularity. The more danceable the more popular a song is. Less acoustic songs seem to be more popular but in general there is not a very significant correlation between popularity and acousticness. The lower the tempo, the more popular the song is, which is surprising since in the previous table, high tempo songs were the most popular. It doesn't seem like there is a significant correlation between popularity and energy which is also surprising since there was a clear difference shown in the previous table.
